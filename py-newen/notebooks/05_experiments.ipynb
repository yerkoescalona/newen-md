# Nine experiments

Each experiment runs a small simulation and draws what it measured over the theory it should match. It then returns a check, a number with the band it must fall in, which this notebook asserts, so it fails if the engine stops reproducing the textbook.

```python
from pathlib import Path

from newen.education import experiments as ex

REF = Path("../../crates/newen-md/tests/gromosXX_references")  # Jupyter started in notebooks/
```

## Where the energy goes

A 2D lattice of argon melts: kinetic and potential energy trade, the total stays put.

```python
r = ex.energy_flow()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## Time step and accuracy

Two argon atoms, run for 8 ps at seven time steps: the energy error grows as dt².

```python
r = ex.time_step_scan()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## The energy landscape

The same pair swings between two turning points on V(r), read from the topology.

```python
r = ex.energy_landscape()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## Thermostats

Berendsen and a Nosé–Hoover chain hold 120 K; only the chain gives canonical fluctuations.

```python
r = ex.thermostat_fingerprint()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## Block averaging

Correlated samples: the error of a mean grows with the block length before it levels off.

```python
r = ex.block_error()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## Maxwell–Boltzmann

Every atom starts at the same speed; collisions spread the speeds into the Maxwell–Boltzmann law.

```python
r = ex.maxwell_boltzmann()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## Liquid structure

A Lennard-Jones liquid: a first shell of neighbours near σ, and g(r) → 1 further out.

```python
r = ex.liquid_structure()
r.plot()
assert all(r.checks), r.checks
r.check()
```

## Water structure

The same measurement on the reference water box: the O–O peak near 0.28 nm.

```python
r = ex.water_structure(
    str(REF / "water_216_box/water_216_box.topo"),
    str(REF / "water_216_box/water_216_box.conf"),
    str(REF / "water_216_nvt/water_216_nvt.in"),
)
r.plot()
assert all(r.checks), r.checks
r.check()
```

## A hot gas meets a cold bath

A 2D gas equilibrated at 240 K is coupled to a 120 K bath. Berendsen brings it to the bath without overshoot, a little more slowly than its τ because the potential energy cools too, and the heat the bath takes, Q, equals what the gas loses; a Nosé–Hoover chain overshoots. Left: the temperature of the three runs, with the bath dashed and the kinetic-only exponentials dotted. Right: the drop of the gas's total energy and Q, on top of each other.

```python
r = ex.heat_bath()
r.plot()
assert all(r.checks), r.checks
r.check()
```
